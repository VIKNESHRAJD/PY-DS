# 1. Square and Cube of Matrix Elements

**AIM**: To create a 3 × 3 matrix using numbers from 1 to 9 and find the square and cube of each element.

```python
import numpy as np

A = np.arange(1, 10).reshape(3, 3)

square = A ** 2
cube = A ** 3

print(square)
```

```python
print(cube)
```

**RESULT:** Thus, the given Python program was successfully executed.

# 2. Row Sum and Column Sum

**AIM**: To create a 4 × 4 matrix using numbers from 1 to 16 and find the row sums and column sums.

```python
import numpy as np

A = np.arange(1, 17).reshape(4, 4)

row_sum = np.sum(A, axis=1)
col_sum = np.sum(A, axis=0)

print(row_sum)
print(col_sum)
```

**RESULT:** Thus, the given Python program was successfully executed.

# 3. 3. Block Matrix

**AIM:** To construct the block matrix \\begin{bmatrix}A&B\\B&A\end{bmatrix} using two given 2 × 2 matrices.

```python
import numpy as np

A = np.array([[1, 2],
              [3, 4]])

B = np.array([[5, 6],
              [7, 8]])

C = np.block([[A, B],
              [B, A]])

print(C)
```

**RESULT:** Thus, the given Python program was successfully executed.

# 4. Matrix Multiplication

**AIM:** To compute the matrix multiplication of two given matrices using NumPy.

# 8. Matrix Slicing

**AIM:** To create a 4 × 4 matrix using numbers from 1 to 16 and extract the first two rows and last two columns

```python
import numpy as np

A = np.arange(1, 17).reshape(4, 4)

extracted = A[0:2, 2:4]

print(extracted)
```

# 9. Block Matrix Using Identity and Zero Matrix

**AIM:** To create the block matrix \begin{bmatrix}0&I\\I&0\end{bmatrix} where I is a 2 × 2 identity matrix

```python
import numpy as np

I = np.eye(2)
Z = np.zeros((2, 2))

A = np.block([[Z, I],
              [I, Z]])

print(I)
```

# 10. Random Matrix Operations and Reshaping

**AIM:** To generate a 4 × 4 random matrix, find its row sums, column sums, square root of elements, and reshape it into 2 × 8.

```python
import numpy as np

A = np.random.rand(4, 4)

row_sum = np.sum(A, axis=1)
col_sum = np.sum(A, axis=0)
sqrt_A = np.sqrt(A)
B = A.reshape(2, 8)

print(A)
```

```python
print("Row sums :",row_sum)
```

```python
print("Column sums :",col_sum)
```

```python
print("Square root of elements :", sqrt_A)
```

```python
print("16 Reshaped (2 x 8 ) :\n", B)
```

**RESULT:** Thus, the given Python program was successfully executed.

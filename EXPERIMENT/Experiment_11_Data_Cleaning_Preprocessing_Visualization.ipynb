# Experiment 11: Data Cleaning, Preprocessing, and Visualization Pipeline Using CSV Datasets

## AIM

To implement a complete data cleaning, preprocessing, and visualization pipeline on CSV datasets using Python for effective exploratory data analysis.

## PROGRAM

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

files = {
    "students": "students.csv",
    "sales": "sales.csv",
    "employees": "employees.csv",
    "health": "health.csv",
    "orders": "orders.csv"
}

def clean_columns(df):
    df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")
    return df

def standardize_text(df):
    for col in df.select_dtypes(include='object').columns:
        df[col] = df[col].astype(str).str.strip().str.lower()
    return df

def handle_missing(df):
    for col in df.columns:
        if df[col].dtype in ['float64', 'int64']:
            df[col] = df[col].fillna(df[col].median())
        else:
            df[col] = df[col].replace("none", np.nan)
            df[col] = df[col].fillna(df[col].mode()[0])
    return df

def remove_duplicates(df):
    return df.drop_duplicates()

def encode_categorical(df):
    for col in df.select_dtypes(include='object').columns:
        df[col] = df[col].astype('category').cat.codes
    return df

def handle_outliers(df):
    numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns
    for col in numeric_cols:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1
        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR
        df[col] = np.where(df[col] < lower, lower,
                           np.where(df[col] > upper, upper, df[col]))
    return df

def feature_engineering(df, name):
    if name == "students":
        if all(col in df.columns for col in ["maths", "physics", "chemistry"]):
            df["total_marks"] = df["maths"] + df["physics"] + df["chemistry"]

    elif name == "sales":
        if "sales" in df.columns and "profit" in df.columns:
            df["profit_ratio"] = df["profit"] / (df["sales"] + 1)

    elif name == "employees":
        if "salary" in df.columns and "experience" in df.columns:
            df["salary_per_exp"] = df["salary"] / (df["experience"] + 1)

    elif name == "health":
        if "weight" in df.columns and "height" in df.columns:
            df["bmi"] = df["weight"] / ((df["height"] / 100) ** 2)

    elif name == "orders":
        if "amount" in df.columns and "deliverytime" in df.columns:
            df["efficiency"] = df["amount"] / (df["deliverytime"] + 1)

    return df

# MAIN PIPELINE
for name, path in files.items():
    print(f"\n========== Processing {name.upper()} ==========")

    df = pd.read_csv(path)

    # Part A & B: Cleaning and Standardization
    df = clean_columns(df)
    df = standardize_text(df)
    df = handle_missing(df)
    df = remove_duplicates(df)
    df = encode_categorical(df)

    # Part C: Outlier Handling
    df = handle_outliers(df)

    # Part D: Feature Engineering
    df = feature_engineering(df, name)

    # Save cleaned file
    df.to_csv(f"cleaned_{name}.csv", index=False)

    print("Shape after cleaning:", df.shape)
    print("Columns:", list(df.columns))

print("\nPIPELINE COMPLETED SUCCESSFULLY")
```

## Program 1: Generator Expression and NumPy Vectorization

### Aim

To calculate the sum of ${i²/(i+1)}$ for i from 1 to 50 using a generator expression and NumPy vectorization.

### Program

```python
import numpy as np

# Generator expression
gen_sum = sum(i*i/(i+1) for i in range(1, 51))

# NumPy vectorized solution
i = np.arange(1, 51)
numpy_sum = np.sum(i*i/(i+1))

print("Generator sum =", gen_sum)
print("NumPy sum     =", numpy_sum)
print("Equal?        =", np.isclose(gen_sum, numpy_sum))
```

## Program 2: Invertibility, Determinant and Matrix Inverse

### Aim

To check whether a random 4 × 4 matrix is invertible and verify its inverse.

### Program

```python
import numpy as np

A = np.random.rand(4, 4)
det_A = np.linalg.det(A)

print("A =\n", A)
print("det(A) =", det_A)

if abs(det_A) > 1e-10:
    A_inv = np.linalg.inv(A)
    product = np.dot(A, A_inv)
    print("A inverse =\n", A_inv)
    print("A @ A_inverse =\n", product)
    print("Verified?", np.allclose(product, np.eye(4)))
else:
    print("Matrix is singular; inverse does not exist.")
```

### Result

Thus, the given Python program was successfully executed.

## Program 4: Eigenvalues and Eigenvectors

### Aim

To calculate the eigenvalues and eigenvectors of a matrix and verify the equation Av = λv.

### Program

```python
import numpy as np

A = np.array([[4, 2],
              [1, 3]], dtype=float)

values, vectors = np.linalg.eig(A)

print("Eigenvalues =", values)
print("Eigenvectors =\n", vectors)

for k in range(len(values)):
    v = vectors[:, k]
    lam = values[k]
    print("\nEigenvalue:", lam)
    print("A v      =", np.dot(A, v))
    print("lambda v =", lam * v)
    print("Verified?", np.allclose(np.dot(A, v), lam * v))
```

### Result

Thus, the given Python program was successfully executed.

## Program 5: Linear System with No Unique Solution

### Aim

To solve a consistent linear system with infinitely many solutions using NumPy and interpret its rank.

### Program

```python
import numpy as np

A = np.array([[1, 1, 1],
              [2, 2, 2]], dtype=float)
b = np.array([3, 6], dtype=float)

x, residuals, rank, singular_values = np.linalg.lstsq(A, b, rcond=None)

print("One NumPy solution =", x)
print("A @ x =", A @ x)
print("b     =", b)
print("Rank(A) =", rank)
print("Residuals =", residuals)

if rank < A.shape[1]:
    print("No unique solution: infinitely many solutions exist.")
```

### Result

Thus, the given Python program was successfully executed.

## Program 6: Rank and Linear Dependence of Rows

### Aim

To find the rank of a matrix and identify the linear dependence between its rows.

### Program

```python
import numpy as np

A = np.array([[1, 2],
              [2, 4]])

rank_A = np.linalg.matrix_rank(A)
print("A =\n", A)
print("Rank(A) =", rank_A)

if rank_A < min(A.shape):
    print("Rows/columns are linearly dependent.")
    print("Here Row 2 = 2 * Row 1.")
```

## Program 7: Non-Commutativity of Matrix Multiplication

### Aim

To compute AB and BA for two random matrices and verify that matrix multiplication is generally non-commutative.

### Program

```python
import numpy as np

A = np.random.rand(3, 3)
B = np.random.rand(3, 3)

AB = np.dot(A, B)
BA = np.dot(B, A)

print("A =\n", A)
print("B =\n", B)
print("AB =\n", AB)
print("BA =\n", BA)
print("AB = BA ?", np.allclose(AB, BA))
```

## Program 8: Block Diagonal Matrix and Its Determinant

### Aim

To construct a block diagonal matrix and verify that its determinant equals the product of the determinants of its diagonal blocks.

### Program

```python
import numpy as np

A = np.array([[1, 2],
              [3, 4]])
B = np.array([[5, 6],
              [7, 8]])

Z = np.zeros((2, 2))
M = np.block([[A, Z],
              [Z, B]])

det_M = np.linalg.det(M)
product = np.linalg.det(A) * np.linalg.det(B)

print("Block matrix M =\n", M)
print("det(M) =", det_M)
print("det(A) det(B) =", product)
print("Verified?", np.isclose(det_M, product))
```

### Result

Thus, the given Python program was successfully executed.

## Program 9: Symmetric Matrix and Orthogonal Eigenvectors

### Aim

To calculate the eigenvalues of a symmetric matrix and verify that its eigenvectors are orthogonal.

### Program

```python
import numpy as np

A = np.array([[2, 1],
              [1, 2]], dtype=float)

values, vectors = np.linalg.eig(A)

v1 = vectors[:, 0]
v2 = vectors[:, 1]
dot_product = np.dot(v1, v2)

print("Eigenvalues =", values)
print("Eigenvectors =\n", vectors)
print("v1 . v2 =", dot_product)
print("Orthogonal?", np.isclose(dot_product, 0))
```

## Program 10: Performance: Python Loops versus NumPy Vectorization

### Aim

To compare the execution time of squaring a 1000 × 1000 matrix using Python loops and NumPy vectorization.

### Program

```python
import numpy as np
import time

A = np.random.rand(1000, 1000)

# NumPy vectorization
start = time.time()
B = A * A
numpy_time = time.time() - start

# Python nested loops / list comprehension
start = time.time()
C = [[A[i, j] * A[i, j] for j in range(1000)]
     for i in range(1000)]
loop_time = time.time() - start

print("NumPy time =", numpy_time, "seconds")
print("Loop time =", loop_time, "seconds")
print("Same result?", np.allclose(B, np.array(C)))

if numpy_time > 0:
    print("Speed-up factor =", loop_time / numpy_time)
```

### Result

Thus, the given Python program was successfully executed.

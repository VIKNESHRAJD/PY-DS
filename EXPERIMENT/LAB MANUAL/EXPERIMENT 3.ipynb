# Experiment 3: List, Tuple, and Set Operations

## 1. Remove Duplicates and Sort in Ascending Order

### Aim

To remove duplicate elements from a list and sort the remaining elements in ascending order.

### Program

```python
lst = [7, 3, 8, 7, 2, 3, 1]
result = sorted(list(set(lst)))
print(result)
```

### Result

Thus, the given Python program was successfully executed.

## 2. Remove Smallest and Largest, Then Reverse

### Aim

To remove the smallest and largest elements from a list and reverse the remaining elements.

### Program

```python
lst = [10, 5, 20, 8, 15, 2]
lst.remove(min(lst))
lst.remove(max(lst))
lst.reverse()
print(lst)
```

### Result

Thus, the given Python program was successfully executed.

## 3. Common Elements of Two Lists

### Aim

To find the common elements between two lists and display them in sorted order.

### Program

```python
list1 = [1, 2, 3, 4, 5]
list2 = [3, 4, 5, 6, 7]
common = sorted(list(set(list1) & set(list2)))
print(common)
```

### Result

Thus, the given Python program was successfully executed.

## 4. Count and Remove All Occurrences of a Word

### Aim

To count the occurrences of a given word in a list and remove all its occurrences.

### Program

```python
words = ["Pele", "Leo", "Messi", "Andres", "Pele"]
word = "Pele"
count = words.count(word)
while word in words:
    words.remove(word)
print("Count:", count)
print("List:", words)
```

### Result

Thus, the given Python program was successfully executed.

## 5. Tuple Index and Count

### Aim

To find the first index and the number of occurrences of a given value in a tuple.

### Program

```python
tuple1 = (2, 5, 10, 5, 8, 10, 1)
value = 10
print("First Index:", tuple1.index(value))
print("Count:", tuple1.count(value))
```

### Result

Thus, the given Python program was successfully executed.

## 6. Copy and Sort a List in Descending Order

### Aim

To create a copy of a list and sort the copied list in descending order without changing the original list.

### Program

```python
lst = [8, 2, 5, 9, 10]
copy_lst = lst.copy()
copy_lst.sort(reverse=True)
print("Sorted:", copy_lst)
print("Original:", lst)
```

### Result

Thus, the given Python program was successfully executed.

### Aim

## 8. Convert List to Set, Add Two Elements, Remove One

### Aim

To convert a list into a set, add two elements, and remove one specified element.

### Program

```python
lst = [1, 2, 2, 3, 4, 4]
s = set(lst)
s.add(5)
s.add(6)
s.remove(2)
print(s)
```

## 9. Common Elements in Three Sets

### Aim

To find the elements that are common to all three sets.

### Program

```python
set1 = {1, 2, 3, 4}
set2 = {2, 3, 5, 6}
set3 = {1, 2, 3}
common = set1.intersection(set2, set3)
print(common)
```

### Result

Thus, the given Python program was successfully executed.

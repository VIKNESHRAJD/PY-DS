## Program 1: Filter Words That Contain a Digit

### Aim

To select words that contain at least one digit.

### Program

```python
words = ["abc", "a1b", "xyz9", "hello "]
result = [w for w in words if any(ch.isdigit() for ch in w)]
print(result)
```

### Result

Thus, the given Python program was successfully executed.

## Program 2: Reverse Words Longer Than Four Characters

### Aim

To reverse words that contain more than four characters.

### Program

```python
sentence = "Python makes programming enjoyable"
words = sentence.split()
result = [w[::-1] for w in words if len(w) > 4]
print(result)
```

### Result

Thus, the given Python program was successfully executed.

## Program 3: Square Even and Cube Odd Positive Numbers

### Aim

To square positive even numbers and cube positive odd numbers.

### Program

```python
nums = [-3, -2, 1, 2, 3, 4]
result = [
    n*n if n % 2 == 0 else n**3
    for n in nums
    if n > 0
]
print(result)
```

### Result

Thus, the given Python program was successfully executed.

## Program 4: Case-Insensitive Palindrome Check

### Aim

To find words that are palindromes regardless of letter case.

### Program

```python
words = ["Level", "python", "Radar", "hello"]
result = [w for w in words if w.lower() == w.lower()[::-1]]
print(result)
```

### Result

Thus, the given Python program was successfully executed.

## Program 5: Uppercase Vowels and Lowercase Consonants

### Aim

To convert vowels to uppercase and consonants to lowercase while ignoring non-alphabetic characters.

### Program

```python
s = "a1B2cDe3"
result = [
    ch.upper() if ch.lower() in "aeiou" else ch.lower()
    for ch in s
    if ch.isalpha()
]
print(result)
```

### Result

Thus, the given Python program was successfully executed.

## Program 6: Words With Matching First and Last Letters

### Aim

To find words whose first and last letters are the same and display their lengths.

### Program

```python
words = ["level", "test", "radar", "python"]
result = [(w, len(w)) for w in words if w[0] == w[-1]]
print(result)
```

### Result

Thus, the given Python program was successfully executed.

## Program 7: Remove Duplicates While Preserving Order

### Aim

To remove duplicate numbers from a list while preserving their original order.

### Program

```python
nums = [1, 2, 2, 3, 1, 4, 3]
result = []
[result.append(n) for n in nums if n not in result]
print(result)
```

### Result

Thus, the given Python program was successfully executed.

## Program 8: Keep Alphabetic Words in Lowercase

### Aim

To keep only alphabetic words and convert them to lowercase.

### Program

```python
words = ["Hello", "World123", "Python", "AI!"]
result = [w.lower() for w in words if w.isalpha()]
print(result)
```

### Result

Thus, the given Python program was successfully executed.

## Program 9: Sort Words by Length in Descending Order

### Aim

To sort the words in a sentence by length from longest to shortest.

### Program

```python
sentence = "Sort these words by length"
result = sorted(sentence.split(), key=len, reverse=True)
print(result)
```

### Result

Thus, the given Python program was successfully executed.

## Program 10: First-and-Last-Letter Abbreviation

### Aim

To create abbreviations using the first and last letters of words with at least two characters.

### Program

```python
words = ["apple", "hi", "a", "banana"]
result = [w[0] + w[-1] for w in words if len(w) >= 2]
print(result)
```

### Result

Thus, the given Python program was successfully executed.

## Program 1: Prime Number Check Using Function

### Aim

To check whether a given number is prime using a function.

### Program

```python
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

n = int(input("Enter a number: "))
print(is_prime(n))
```

### Result

Thus, the given Python program was successfully executed.

## Program 2: Fibonacci Numbers Less Than n

### Aim

To generate all Fibonacci numbers less than a given number n.

### Program

```python
n = int(input("Enter n: "))
a, b = 0, 1
while a < n:
    print(a, end=" ")
    a, b = b, a + b
print()
```

### Result

Thus, the given Python program was successfully executed.

## Program 4: Character Type Count

### Aim

To count uppercase letters, lowercase letters, digits, and special characters in a string.

### Program

```python
s = input("Enter a string: ")
upper = lower = digit = special = 0

for ch in s:
    if ch.isupper():
        upper += 1
    elif ch.islower():
        lower += 1
    elif ch.isdigit():
        digit += 1
    else:
        special += 1

print(upper, lower, digit, special)
```

### Result

Thus, the given Python program was successfully executed.

## Program 5: Merge Two Lists Without Duplicates

### Aim

To merge two lists, remove duplicates, and preserve the original order.

### Program

```python
l1 = list(map(int, input("Enter first list: ").split()))
l2 = list(map(int, input("Enter second list: ").split()))

result = []
for x in l1 + l2:
    if x not in result:
        result.append(x)

print(result)
```

## Program 7: GCD Using Function

### Aim

To find the greatest common divisor of two numbers using a function.

### Program

```python
def gcd(a, b):
    while b:
        a, b = b, a % b
    return a

print(gcd(12, 18))
```

### Result

Thus, the given Python program was successfully executed.

## Program 11: Armstrong Numbers in a Range

### Aim

To find all Armstrong numbers from 1 to 999.

### Program

```python
for num in range(1, 1000):
    order = len(str(num))
    if num == sum(int(d) ** order for d in str(num)):
        print(num)
```

### Result

Thus, the given Python program was successfully executed.

## Program 14: Decimal to Binary, Octal, and Hexadecimal

### Aim

To convert a decimal number to binary, octal, and hexadecimal without built-in conversion functions.

### Program

```python
n = int(input("Enter a decimal number: "))

def convert(n, base):
    res = ""
    digits = "0123456789ABCDEF"
    while n > 0:
        res = digits[n % base] + res
        n //= base
    return res if res else "0"

print(convert(n, 2), convert(n, 8), convert(n, 16))
```

### Result

Thus, the given Python program was successfully executed.

## Program 17: Password Validation

### Aim

To validate a password with at least 8 characters, one uppercase letter, one lowercase letter, and one digit.

### Program

```python
pwd = input("Enter password: ")
valid = (
    len(pwd) >= 8 and
    any(c.isupper() for c in pwd) and
    any(c.islower() for c in pwd) and
    any(c.isdigit() for c in pwd)
)
print(valid)
```

## Program 19: Menu-Driven Calculator Using Functions

### Aim

To implement addition, subtraction, multiplication, and division using functions.

### Program

```python
def add(a, b):
    return a + b

def sub(a, b):
    return a - b

def mul(a, b):
    return a * b

def div(a, b):
    return a / b if b != 0 else None

a = float(input("Enter first number: "))
b = float(input("Enter second number: "))
choice = input("Choose +, -, *, /: ")

if choice == "+":
    print(add(a, b))
elif choice == "-":
    print(sub(a, b))
elif choice == "*":
    print(mul(a, b))
elif choice == "/":
    print(div(a, b))
else:
    print("Invalid choice")
```

### Result

Thus, the given Python program was successfully executed.

## Program 20: Remove Duplicate Words

### Aim

To remove duplicate words from a sentence while maintaining the original order.

### Program

```python
sentence = input("Enter a sentence: ").split()
result = []
for word in sentence:
    if word not in result:
        result.append(word)
print(" ".join(result))
```

### Result

Thus, the given Python program was successfully executed.

## Program 1: Random Sensor Readings and Normalization

### Aim

To generate 1000 random sensor readings and normalize them using the mean and standard deviation.

### Program

```python
import numpy as np

# Generate 1000 random numbers
data = np.fromiter((np.random.rand() for _ in range(1000)), dtype=float)

# Compute mean and standard deviation of original data
mean = np.mean(data)
std = np.std(data)

# Normalize the data
normalized = (data - mean) / std

# Print results
print("Original Mean =", mean)
print("Original Standard Deviation =", std)
print("Normalized Data =", normalized)

# Check mean and standard deviation of normalized data
print("Normalized Mean =", np.mean(normalized))
print("Normalized Standard Deviation =", np.std(normalized))
print(np.isclose(np.mean(normalized), 0))
print(round(np.mean(normalized), 10))
```

## Program 3: Monte-Carlo Estimation of Pi

### Aim

To estimate the value of pi by generating random points and counting those inside a unit circle.

### Program

```python
import numpy as np

N = 10000
points = ((np.random.rand(), np.random.rand()) for _ in range(N))
inside = sum(1 for x, y in points if x*x + y*y <= 1)
pi_est = 4 * inside / N

print("Estimated Pi:", pi_est)
```

Thus, the given Python program was successfully executed.

## Program 4: Stability of Random 3 × 3 Matrices

### Aim

To generate 10 random 3 × 3 matrices and display matrices whose eigenvalues all have magnitude less than 1.

### Program

```python
import numpy as np

# Generate 10 random 3x3 matrices
matrices = [np.random.randn(3, 3) for _ in range(10)]

# Select only stable matrices
stable = (A for A in matrices if np.all(np.abs(np.linalg.eigvals(A)) < 1))

# Print each stable matrix
for s in stable:
    print("Stable Matrix:\n", s)
```

## Program 5: Sliding Windows and Covariance

### Aim

To create length-10 sliding windows from 200 random values and calculate the covariance/variance for each window.

### Program

```python
import numpy as np

data = np.random.rand(200)
windows = (data[i:i+10] for i in range(len(data)-9))
covariances = [np.cov(w) for w in windows]

print(covariances)
```

## Program 8: Principal Component Analysis (PCA)

### Aim

To calculate the covariance matrix of a 100 × 5 dataset and project the centered data onto the first principal component.

### Program

```python
import numpy as np

X = np.random.rand(100, 5)
X_centered = X - np.mean(X, axis=0)
cov = np.cov(X_centered, rowvar=False)
eigvals, eigvecs = np.linalg.eig(cov)
pc1 = eigvecs[:, np.argmax(eigvals)]
projection = X_centered @ pc1

print(projection)
```

## Program 9: Most Influential Node from a Dominant Eigenvector

### Aim

To generate a 6 × 6 adjacency matrix and identify the node with the largest absolute component in the selected dominant eigenvector.

### Program

```python
import numpy as np

A = np.random.randint(0, 2, (6, 6))
eigvals, eigvecs = np.linalg.eig(A)
dominant = eigvecs[:, np.argmax(eigvals)]
node = max((i for i in range(len(dominant))),
           key=lambda i: abs(dominant[i]))

print("Most influential node:", node)
```

# Program 10: Broadcasting – Robot Spare-Part Billing

### Aim

To apply NumPy broadcasting to calculate final billing values after discounts and taxes and identify the group with maximum revenue.



### Program

```python
import numpy as np

S = np.array([[320, 360, 400],
              [280, 310, 350],
              [450, 480, 520],
              [240, 270, 300]])

d = np.array([20, 25, 30])
t = np.array([[0.05], [0.08], [0.10], [0.06]])

final = (S - d) * (1 + t)
revenue = final.sum(axis=1)
best = np.argmax(revenue)

print("Shape:", final.shape)
print(final)
print("Revenue:", revenue)
print("Maximum group:", best + 1)
```

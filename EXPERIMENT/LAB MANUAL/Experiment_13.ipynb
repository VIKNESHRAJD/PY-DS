## Program 1: Normal Distribution

**AIM:**

To generate exam scores from a normal distribution with mean 65 and standard deviation 12, estimate the probability of scoring above 80, and visualize the distribution.

**PROGRAM:**

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.normal(65, 12, 1000)
prob = np.mean(data > 80)
print("Probability of score >80:", prob)

plt.hist(data, bins=30)
plt.title("Normal Distribution of Exam Scores")
plt.xlabel("Score")
plt.ylabel("Frequency")
plt.show()
```

**RESULT:**

The program estimates the proportion of scores above 80 and displays their distribution. The numerical estimate varies between runs.

## Program 2: Standard Normal Distribution

**AIM:**

To convert normally distributed data into z-scores and verify that the transformed data has mean approximately 0 and standard deviation approximately 1.

**PROGRAM:**

```python
import numpy as np

data = np.random.normal(50, 10, 1000)
z = (data - np.mean(data)) / np.std(data)

print("Mean of Z-scores:", np.mean(z))
print("Std of Z-scores:", np.std(z))
```

**RESULT:**

The data is standardized; its z-scores have mean approximately 0 and standard deviation 1.

## Program 3: Binomial Distribution

**AIM:**

To simulate 1000 experiments of tossing a fair coin 10 times and determine the most frequent number of heads.

**PROGRAM:**

```python
import numpy as np

data = np.random.binomial(10, 0.5, 1000)
values, counts = np.unique(data, return_counts=True)
mode_heads = values[np.argmax(counts)]

print("Most frequent number of heads:", mode_heads)
```

**RESULT:**

The simulation identifies the number of heads that occurs most frequently across the 1000 trials. The result may vary between runs.

## Program 5: Uniform Distribution

**AIM:**

To estimate the value of pi using Monte Carlo simulation with uniformly distributed points in a square.

**PROGRAM:**

```python
import numpy as np

x = np.random.uniform(-1, 1, 10000)
y = np.random.uniform(-1, 1, 10000)
inside = x**2 + y**2 <= 1
pi_estimate = 4 * np.mean(inside)

print("Estimated Pi:", pi_estimate)
```

**RESULT:**

The Monte Carlo simulation estimates pi by using the proportion of points inside the unit circle. The estimate varies between runs.

## Program 9: Student's t Distribution

**AIM:**

To compare the probability that absolute values exceed 2 for a Student's t distribution with 10 degrees of freedom and a standard normal distribution.

**PROGRAM:**

```python
import numpy as np
from scipy.stats import t

data_t = t.rvs(df=10, size=1000)
data_n = np.random.normal(0, 1, 1000)

prob_t = np.mean(np.abs(data_t) > 2)
prob_n = np.mean(np.abs(data_n) > 2)

print("P(|T|>2):", prob_t)
print("P(|N|>2):", prob_n)
```

**RESULT:**

The simulation typically shows a larger tail probability for the Student's t distribution than for the standard normal distribution. Results vary between runs.

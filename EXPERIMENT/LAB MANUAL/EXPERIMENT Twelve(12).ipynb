# Experiment 12: Data Visualization Using Line, Bar, Scatter, and Pie Charts

## AIM

To visualize data graphically using line plots, bar charts, scatter plots, and pie charts in Python for effective analysis and interpretation of datasets.

## PROGRAM 1: Diagrams Shown in the Lab Record

```python
import numpy as np
import matplotlib.pyplot as plt

# Line Plot
x = np.arange(0, 7)
m = 250 / 6
y = m * x
plt.plot(x, y, color='red')
plt.grid(True)
plt.title('Line Plot')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.show()
```

```python
# Pie Chart
labels = ['Children 0-18', 'Adults 19-25', 'Adults 26-34', 'Adults 35-54', 'Adults 55-64', 'Seniors 65+']
sizes = [23.59, 8.71, 12.47, 25.52, 13.18, 16.53]
explode = [0, 0, 0, 0.08, 0.10, 0]
plt.figure()
plt.pie(sizes, labels=labels, autopct='%1.2f%%', explode=explode, startangle=90)
plt.axis('equal')
plt.show()
```

```python
# Donut Chart
labels = ['apple', 'banana', 'cherry', 'durian', 'elderberries', 'figs', 'grapes']
sizes = [32, 20, 15, 10, 10, 8, 5]
plt.figure(figsize=(8, 8))
plt.pie(sizes, labels=labels, autopct='%1.1f%%', wedgeprops=dict(width=0.4))
plt.title('Fruit Distribution (Donut Chart)')
plt.axis('equal')
plt.show()
```

```python
#Nested Donut Chart
outer_labels = ['fruit', 'vegetable']
outer_sizes = [40, 60]
inner_labels = ['apple', 'banana', 'cherry', 'durian', 'onion', 'tomato', 'potato']
inner_sizes = [10, 10, 10, 10, 10, 15, 35]
fig, ax = plt.subplots(figsize=(8, 8))
ax.pie(outer_sizes, labels=outer_labels, radius=1, startangle=90, wedgeprops=dict(width=0.3, edgecolor='white'))
ax.pie(inner_sizes, labels=inner_labels, radius=0.7, startangle=90, labeldistance=0.68, wedgeprops=dict(width=0.3, edgecolor='white'))
ax.set(aspect='equal')
plt.title('Nested Donut Chart')
plt.show()
```

## PROGRAM 2: Generate 100 Random Integers and Draw Four Plots

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.randint(1, 100, 100)
x = np.arange(1, 101)

plt.figure()
plt.plot(x, data, label='Random Data')
plt.title('Line Plot of 100 Random Integers')
plt.xlabel('Index')
plt.ylabel('Value')
plt.legend()
plt.show()
```

```python

plt.figure()
plt.scatter(x, data, label='Random Data')
plt.title('Scatter Plot of 100 Random Integers')
plt.xlabel('Index')
plt.ylabel('Value')
plt.legend()
plt.show()
```

```python
plt.figure()
plt.bar(x, data, label='Random Data')
plt.title('Bar Plot of 100 Random Integers')
plt.xlabel('Index')
plt.ylabel('Value')
plt.legend()
plt.show()
```

```python
plt.figure()
plt.barh(x, data, label='Random Data')
plt.title('Horizontal Bar Plot of 100 Random Integers')
plt.xlabel('Value')
plt.ylabel('Index')
plt.legend()
plt.show()
```

## PROGRAM 3: Histograms of Four Probability Distributions

```python
import numpy as np
import matplotlib.pyplot as plt

binomial = np.random.binomial(n=10, p=0.5, size=1000)
plt.figure()
plt.hist(binomial)
plt.title('Binomial Distribution')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()
```

```python

expo = np.random.exponential(scale=1.0, size=1000)
plt.figure()
plt.hist(expo)
plt.title('Exponential Distribution')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()
```

```python
normal = np.random.normal(loc=0, scale=1, size=1000)
plt.figure()
plt.hist(normal)
plt.title('Normal Distribution')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()
```

```python
uniform = np.random.uniform(low=0, high=1, size=1000)
plt.figure()
plt.hist(uniform)
plt.title('Uniform Distribution')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()
```

## PROGRAM 4: Plot the Given Data Using Suitable Diagrams

```python
import matplotlib.pyplot as plt

x1 = [1, 2, 3, 4, 5, 6, 7]
y1 = [10, 15, 13, 18, 20, 22, 25]
x2 = ['A', 'B', 'C', 'D', 'E']
y2 = [25, 40, 30, 35, 20]
x3 = [5, 10, 15, 20, 25, 30, 35]
y3 = [12, 18, 20, 28, 30, 33, 40]
labels4 = ['Rent', 'Food', 'Transport', 'Savings']
sizes4 = [40, 25, 15, 20]
x5 = [1, 2, 3, 4, 5]
y5_1 = [50, 55, 60, 65, 70]
y5_2 = [45, 50, 52, 58, 62]

plt.figure()
plt.plot(x1, y1, marker='o', label='Trend')
plt.title('Line Plot')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.show()
```

```python
plt.figure()
plt.bar(x2, y2, label='Categories')
plt.title('Bar Plot')
plt.xlabel('Category')
plt.ylabel('Values')
plt.legend()
plt.show()
```

```python
plt.figure()
plt.scatter(x3, y3, label='Data Points')
plt.title('Scatter Plot')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.show()
```

```python
plt.figure()
plt.pie(sizes4, labels=labels4, autopct='%1.1f%%')
plt.title('Expense Distribution')
plt.show()
```

```python
plt.figure()
plt.plot(x5, y5_1, marker='o', label='Series 1')
plt.plot(x5, y5_2, marker='s', label='Series 2')
plt.title('Multiple Line Plot')
plt.xlabel('X')
plt.ylabel('Values')
plt.legend()
plt.show()
```

## Program 10: Armstrong Number Check

### AIM

To check whether a nonnegative integer is an Armstrong number.

### PROGRAM

```python
n = int(input("Enter a number: "))
digits = list(map(int, str(n)))
power = len(digits)
if sum(d ** power for d in digits) == n:
    print("Armstrong number")
else:
    print("Not an Armstrong number")
```

### RESULT

Thus, the given Python program was successfully executed.

## Program 12: First n Fibonacci Numbers

### AIM

To print the first n terms of the Fibonacci sequence.

### PROGRAM

```python
n = int(input("Enter n: "))
a, b = 0, 1
for _ in range(n):
    print(a, end=" ")
    a, b = b, a + b
print()
```

### RESULT

Thus, the given Python program was successfully executed.

## Program 14: Prime Numbers up to n

### AIM

To print every prime number from 2 through n.

### PROGRAM

```python
n = int(input("Enter n: "))
for num in range(2, n + 1):
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            break
    else:
        print(num, end=" ")
print()
```

### RESULT

Thus, the given Python program was successfully executed.

## Program 18: Word Frequency Using a Dictionary

### AIM

To count how many times each word appears in a sentence.

### PROGRAM

```python
sentence = input("Enter a sentence: ")
freq = {}
for word in sentence.split():
    freq[word] = freq.get(word, 0) + 1
print(freq)
```

### RESULT

Thus, the given Python program was successfully executed.

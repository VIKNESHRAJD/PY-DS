# Case Study 3: Sales and Customer Analysis

Analyze product sales, prices, quantities, revenue, and customer information. Use
grouping and aggregation to identify sales patterns and summarize business
performance.

```python
import pandas as pd
```

```python
data = {
    "Order_ID": [1001, 1002, 1003, 1004, 1005, 1006, 1007, 1008, 1009, 1010],
    "Customer_Name": ["Vaisu", "Vikki", "Vendhan", "Siri", "Priya","Perumal", "Atena", "Leo", "Vaisu", "Vikki"],
    "Region": ["South", "North", "South", "East", "West", "North", "East", "West", "South", "North"],
    "Product": ["Laptop", "Mobile", "Tablet", "Laptop", "Mobile", "Tablet", "Laptop", "Mobile", "Tablet", "Laptop"],
    "Category": ["Electronics", "Electronics", "Electronics", "Electronics",
                 "Electronics", "Electronics", "Electronics", "Electronics",
                 "Electronics", "Electronics"],
    "Price": [55000, 20000, 30000, 50000, 18000, 32000, 60000, 22000, 28000, 58000],
    "Quantity": [2, 5, 2, 1, 3, 4, 2, 6, 3, 2],
    "Customer_Type": ["New", "Returning", "New", "Returning", "New", "Returning", "New", "Returning", "Returning", "New"]
}

df = pd.DataFrame(data)

df
```

## Calculate Revenue

Revenue is calculated using price and quantity.

$$
\text{Revenue}
=
\text{Price} \times \text{Quantity}
$$

```python
df["Revenue"] = df["Price"] * df["Quantity"]
df
```

## Product-wise Sales Analysis

Group the data by product and calculate:

- Total quantity sold
- Total revenue
- Average price

```python
product_analysis = df.groupby("Product").agg(
    Total_Quantity=("Quantity", "sum"),
    Total_Revenue=("Revenue", "sum"),
    Average_Price=("Price", "mean")
)

product_analysis
```

## Customer-wise Sales Analysis

Group the data by customer and calculate:

- Number of orders
- Total quantity purchased
- Total revenue

```python
customer_analysis = df.groupby("Customer_Name").agg(
    Number_of_Orders=("Order_ID", "count"),
    Total_Quantity=("Quantity", "sum"),
    Total_Revenue=("Revenue", "sum")
)

customer_analysis
```

## Region-wise Sales Analysis

Group the data by region and calculate:

- Number of orders
- Total quantity sold
- Total revenue

```python
region_analysis = df.groupby("Region").agg(
    Number_of_Orders=("Order_ID", "count"),
    Total_Quantity=("Quantity", "sum"),
    Total_Revenue=("Revenue", "sum")
)

region_analysis
```

##  Identifying Sales Patterns

Sort the product and customer analysis by total revenue
to identify the highest revenue-generating products and customers.

```python
product_analysis.sort_values(by="Total_Revenue", ascending=False)
```

```python
customer_analysis.sort_values(by="Total_Revenue", ascending=False)
```

## Business Performance Summary

Calculate the total quantity sold and total revenue
to summarize overall business performance.

* **Total quantity sold** → How many products were sold overall.
* **Total revenue** → How much money was generated from sales.
* **Best-selling product** → Which product sold the highest quantity.
* **Highest revenue product** → Which product generated the most revenue.
* **Top customer** → Which customer generated the most revenue.
* **Best-performing region** → Which region generated the most revenue.

```python
business_summary = df.agg(Total_Quantity_Sold=("Quantity", "sum"), Total_Revenue=("Revenue", "sum"))
business_summary
```

```python
business_summary = pd.DataFrame({
    "Metric": ["Total Quantity Sold", "Total Revenue"],
    "Value": [df["Quantity"].sum(), df["Revenue"].sum()]
})

business_summary
```

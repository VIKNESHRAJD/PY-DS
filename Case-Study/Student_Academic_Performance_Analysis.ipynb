##  Mathematical Formulation
### 1. Mean (Average Marks)
$$
\bar{x} = \frac{1}{n}\sum_{i=1}^{n} x_i
$$

### 2. Standard Deviation
$$
\sigma = \sqrt{\frac{1}{n}\sum_{i=1}^{n}(x_i - \bar{x})^2}
$$

### 3. Total Marks
$$
\text{Total} = \sum_{k=1}^{m} \text{Subject}_k
$$

### 4. Percentage
$$
\text{Percentage} = \left(\frac{\text{Total Marks Obtained}}{\text{Total Maximum Marks}}\right) \times 100
$$

### 5. Attendance Ratio
$$
\text{Attendance Ratio} = \frac{\text{Classes Attended}}{\text{Total Classes}} \times 100
$$

## **Data Cleaning**

> Data cleaning is the process of identifying and correcting missing,
incorrect, or inconsistent data.

> First, we check whether the dataset contains missing values.

## Handling Missing Values


We first detect missing values and then replace them with
appropriate values.

For numerical data, mean or median can be used depending
on the characteristics of the data.

```python
df.isnull().sum()
```

```python
df["Attendance_%"] = df["Attendance_%"].fillna(df["Attendance_%"].mean())

df["Mathematics"] = df["Mathematics"].fillna(df["Mathematics"].mean())

df["Chemistry"] = df["Chemistry"].fillna(df["Chemistry"].mean())

df["German"] = df["German"].fillna(df["German"].mean())

df
```

```python
# Check missing values AFTER cleaning
df.isnull().sum()
```

Since the dataset does not contain missing values or duplicate records,
no major cleaning operation is required.


## **Indexing and Selection**

The `Student_ID` column uniquely identifies each student.

It can be used as the DataFrame index to make student-level analysis
more convenient.


Pandas provides two important indexing methods:

- `loc[]` → label-based indexing
- `iloc[]` → position-based indexing

`iloc[]` works like normal Python positional indexing,
while `loc[]` selects data using labels.

```python
df_indexed = df.set_index("Student_ID")

df_indexed
```

Now Student_ID becomes the row label.

```python
df_indexed.loc[102]
```

This retrieves the student whose Student_ID is 102.

```python
df_indexed.iloc[1]
```

This retrieves the second row because Python positions start from 0.

## Calculating Average Marks

The average mark of each student can be calculated using the arithmetic
mean.

### Academic Performance Calculation

Each student has marks in five subjects:

1. Mathematics
2. Physics
3. Chemistry
4. Social Science
5. German

The total marks are calculated as:

$$
\text{Total Marks}
=
M + P + C + SS + G
$$

The average mark is:

$$
\text{Average Marks}
=
\frac{\text{Total Marks}}{5}
$$

```python
subjects = ["Mathematics", "Physics", "Chemistry", "Social_Science", "German"]

df["Total_Marks"] = df[subjects].sum(axis=1)

df["Average_Marks"] = df["Total_Marks"] / len(subjects)

df
```

>> axis=1 means the calculation is performed row-wise.

Therefore, the five subject marks of each student are added together.

## Filtering Students

Boolean conditions can be used to select specific students.

For example, students with an average mark of at least 80
can be identified using a filtering condition.

```python
Top_scorers = df[df["Average_Marks"] >= 80]

Top_scorers[["Student_ID", "Name", "Average_Marks", "Grade"]]
```

## Grouping and Aggregation

The `groupby()` function is used to divide the data into groups based
on a categorical variable.

- Number of students
- Average attendance
- Average marks

```python
gender_analysis = df.groupby("Gender").agg(
    Student_Count=("Student_ID", "count"),
    Average_Attendance=("Attendance_%", "mean"),
    Average_Marks=("Average_Marks", "mean")
)

gender_analysis
```

**Interpretation**

> This allows us to compare academic and attendance statistics between the M and F groups without manually calculating each group.

## Grouping Students by Grade

Students are grouped according to their final grade.

The analysis calculates:

- Number of students
- Average marks
- Average attendance

```python
grade_analysis = df.groupby("Grade").agg(
    Student_Count=("Student_ID", "count"),
    Average_Marks=("Average_Marks", "mean"),
    Average_Attendance=("Attendance_%", "mean")
)

grade_analysis
```

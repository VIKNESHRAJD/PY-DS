### BMI

Body Mass Index (BMI) is calculated as:

$$
BMI = \frac{\text{Weight (kg)}}{\text{Height (m)}^2}
$$

BMI is used as a general measure of body weight relative to height.

## Creating the Healthcare Dataset

The dataset contains information about:

- Patient age
- Gender
- BMI
- Blood pressure
- Disease status

Some values are intentionally missing so that data cleaning
can be demonstrated.

```python
data = {
    "Patient_ID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Patient_Name": ["Vaisu", "Vikki", "Vendhan", "Siri", "Priya", "Perumal", "Atena","Leo"],
    "Age": [21, 21, 45, 52, np.nan, 41, 60, 29],
    "Gender": ["F", "M", "M", "F", "F", "M", "F", "M"],
    "BMI": [22.5, 27.8, 31.2, np.nan, 24.6, 29.5, 33.1, 23.4],
    "Blood_Pressure": [118, 125, 140, 135, 120, np.nan, 150, 115],
    "Disease_Status": ["No", "No", "Yes", "Yes", "No", "Yes", "Yes", "No"]
}

df = pd.DataFrame(data)

df
```

##  Handling Missing Values

The missing values in the numerical columns are replaced using
the mean of the corresponding column.

Mean imputation is suitable for numerical data when the data
is reasonably suitable for using the mean as a summary.

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())

df["BMI"] = df["BMI"].fillna(df["BMI"].mean())

df["Blood_Pressure"] = df["Blood_Pressure"].fillna(df["Blood_Pressure"].mean())

df
```

```python
#Check missing values again
df.isnull().sum()
```

## Identifying Patterns

The healthcare data can be grouped according to disease status.

The average age, BMI, and blood pressure are calculated for
patients with and without the disease.

This helps identify differences between the two groups.

```python
disease_analysis = df.groupby("Disease_Status").agg(
    Patient_Count=("Patient_ID", "count"),
    Average_Age=("Age", "mean"),
    Average_BMI=("BMI", "mean"),
    Average_Blood_Pressure=("Blood_Pressure", "mean")
)

disease_analysis
```

##  Probability Distribution

A probability distribution describes how probability is
associated with possible values of a random variable.

For the healthcare dataset, disease status can be treated
as a binary outcome:

- `Yes` → disease present
- `No` → disease absent

The probability of each disease status is calculated from
the observed frequency.

For example:

$$
P(\text{Disease = Yes})
=
\frac{\text{Number of Yes patients}}
{\text{Total number of patients}}
$$

and

$$
P(\text{Disease = No})
=
\frac{\text{Number of No patients}}
{\text{Total number of patients}}
$$

```python
disease_probability = (df["Disease_Status"].value_counts(normalize=True))

disease_probability
```

> Normalize = True

 Converts the frequencies into proportions.

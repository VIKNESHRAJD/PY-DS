# Case Study 4: Employee Salary and Performance Analysis



Analyze employee salary, experience, department, performance rating,
and job role using Pandas.

The analysis includes:

- Data cleaning
- Grouping
- Aggregation
- Descriptive statistics
- Relationship between experience, salary, and performance.

# Import Libraries

```python
import pandas as pd
import numpy as np
```

## Creating the Employee Dataset

The dataset contains employee salary, experience, department,
performance rating, and job role.

Some values are missing so that data cleaning can be performed.

```python
data = {
    "Employee_ID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Employee_Name": ["Vaisu", "Vikki", "Priya", "Meena", "Leo", "Siri", "Vendhan", "Atena"],
    "Experience_Years": [2, 5, 3, np.nan, 8, 4, 10, 1],
    "Department": ["HR", "IT", "Finance", "IT", "HR", "Finance", "IT", "HR"],
    "Salary": [450000, 650000, 480000, 70000, 85000, np.nan, 95000, 32000],
    "Performance_Rating": [3, 4, 3, 5, 5, 4, 5, 3],
    "Job_Role": ["HR Executive", "Software Engineer", "Accountant",
                 "Software Engineer", "HR Manager", "Financial Analyst",
                 "Senior Engineer", "HR Executive"]
}

df = pd.DataFrame(data)

df
```

## Data Cleaning

First, missing values are identified.

Missing numerical values are handled using the mean of the
corresponding column.

```python
df.isnull().sum()
```

## Filling Missing Value

```python
df["Experience_Years"] = df["Experience_Years"].fillna(df["Experience_Years"].mean())

df["Salary"] = df["Salary"].fillna(df["Salary"].mean())

df
```

```python
# Checking again
df.isnull().sum()
```

## Department-wise Analysis

Employees are grouped according to their department.

The following values are calculated:

- Number of employees
- Average salary
- Average experience
- Average performance rating

```python
department_analysis = df.groupby("Department").agg(
    Employee_Count=("Employee_ID", "count"),
    Average_Salary=("Salary", "mean"),
    Average_Experience=("Experience_Years", "mean"),
    Average_Performance=("Performance_Rating", "mean")
)

department_analysis
```

## Grouping by Job Role

Employees are grouped according to their job role.

The average salary and performance rating are calculated
for each job role.

```python
role_analysis = df.groupby("Job_Role").agg(
    Employee_Count=("Employee_ID", "count"),
    Average_Salary=("Salary", "mean"),
    Average_Performance=("Performance_Rating", "mean")
)

role_analysis
```

## Salary and Performance Aggregation

Aggregation is used to calculate overall salary and
performance measures.

```python
salary_performance = df.agg(
    Average_Salary=("Salary", "mean"),
    Maximum_Salary=("Salary", "max"),
    Minimum_Salary=("Salary", "min"),
    Average_Performance=("Performance_Rating", "mean")
)

salary_performance
```

## Descriptive Statistical Analysis

Descriptive statistics are calculated for:

- Experience
- Salary
- Performance rating

The `describe()` function provides count, mean, standard deviation,
minimum, quartiles, and maximum.

```python
df[["Experience_Years", "Salary", "Performance_Rating"]].describe()
```

## Relationship Between Experience, Salary and Performance

The correlation between experience, salary, and performance
is calculated to study their relationship.

The correlation coefficient ranges from -1 to +1.

- Positive value → positive relationship
- Negative value → negative relationship
- Value close to 0 → weak or no linear relationship

## Correlation Between Experience, Salary and Performance

The Pearson correlation coefficient between two variables is given by:

$$
r_{XY}
=
\frac{
\sum_{i=1}^{n}(X_i-\bar{X})(Y_i-\bar{Y})
}{
\sqrt{
\sum_{i=1}^{n}(X_i-\bar{X})^2
\sum_{i=1}^{n}(Y_i-\bar{Y})^2
}
}
$$

The correlation matrix calculates the correlation between:

- **Experience and Salary**
- **Experience and Performance**
- **Salary and Performance**

The value of $r$ ranges from $-1$ to $+1$.

$$
-1 \leq r \leq +1
$$

A positive value indicates a positive relationship, while a negative value
indicates a negative relationship.

```python
relationship = df[["Experience_Years", "Salary", "Performance_Rating"]].corr()
relationship
```

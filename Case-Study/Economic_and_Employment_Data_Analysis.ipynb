- Income
- Employment

The analysis includes data cleaning, Pandas operations, descriptive statistics,

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## 1. Create the Dataset

The dataset represents economic and employment indicators for different countries.

```python
data = {
    "Country": ["India", "USA", "Germany", "UK", "Japan", "Canada", "Australia", "France", "Brazil", "China"],

    "Income_USD": [2500, 76000, 95000, 46000, 42000, 55000, 65000, 45000, 11000, 13000],

    "Employment_Rate": [57.0, 60.2, 77.2, 75.1, 78.5, 61.8, 77.0, 68.5, 56.0, 67.5],

    "Inflation_Rate": [5.4, 3.2, 2.1, 3.0, 2.8, 2.9, 3.4, 2.7, 4.6, 2.1],

    "Education_Years": [6.7, 13.7, 14.1, 13.6, 13.5, 13.9, 13.2, 13.4, 8.6, 10.8],

    "Population_Million": [1428, 335, 84, 68, 124, 40, 27, 68, 216, 1410]
}

df = pd.DataFrame(data)
df
```

## Inspect the Data

Before performing analysis, the dataset is inspected to understand its structure,
data types, and missing values.

```python
df.isnull().sum()
```

## Data Cleaning

Missing values are checked before analysis.

If missing numerical values are present, they can be replaced using the mean
of the corresponding column.

```python
numeric_columns = ["Income_USD", "Employment_Rate", "Inflation_Rate", "Education_Years", "Population_Million"]

for column in numeric_columns:
    df[column] = df[column].fillna(df[column].mean())

df.isnull().sum()
```

## Pandas Operations

The income, employment rate, inflation rate, education level, and population
columns are selected for economic analysis.

```python
economic_data = df[["Country","Income_USD", "Employment_Rate", "Inflation_Rate", "Education_Years", "Population_Million"]]
economic_data
```

## Descriptive Statistics

Descriptive statistics are used to understand the central tendency and

```python
df[["Income_USD", "Employment_Rate", "Inflation_Rate", "Education_Years", "Population_Million"]].describe()
```

## Grouping and Aggregation

Countries are grouped according to their income level.

The income classification is created using the income value.

```python
def income_group(income):
    if income < 10000:
        return "Low Income"
    elif income < 40000:
        return "Middle Income"
    else:
        return "High Income"

df["Income_Group"] = df["Income_USD"].apply(income_group)
df
```

```python
income_analysis = df.groupby("Income_Group").agg(
    Number_of_Countries=("Country", "count"),
    Average_Income=("Income_USD", "mean"),
    Average_Employment=("Employment_Rate", "mean"),
    Average_Inflation=("Inflation_Rate", "mean"),
    Average_Education=("Education_Years", "mean"),
    Total_Population=("Population_Million", "sum")
)

income_analysis
```

## Probability of High Employment

A country is considered to have a high employment rate when its employment
rate is greater than 70%.

The probability is calculated as:

$$
P(\text{High Employment})
=
\frac{\text{Number of countries with employment rate > 70}}
{\text{Total number of countries}}
$$

```python
high_employment = (df["Employment_Rate"] > 70).sum()
total_countries = len(df)

probability_high_employment = high_employment / total_countries

print("Probability of high employment:",
      probability_high_employment)
```

```python
# Lab 9 — Programs 1 to 9 in a single Jupyter cell
# Requirements: numpy, pandas, scikit-learn, networkx (used where applicable)

import numpy as np
import pandas as pd
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
import networkx as nx

print("--- Lab9 Program 1: Random Sensor Readings and Normalization ---")
def lab1():
    data = np.random.rand(1000)
    mean = np.mean(data)
    std = np.std(data)
    normalized = (data - mean) / std
    print("Original Mean =", mean)
    print("Original Std =", std)
    print("Normalized mean ~ 0?", np.isclose(np.mean(normalized), 0.0))
    print()

lab1()

print("--- Lab9 Program 2: Sum of Squares of Non-zero Matrix Elements ---")
def lab2():
    A = np.random.randint(-5, 5, size=(50,50))
    nonzero = A[A != 0]
    energy = np.sum(nonzero**2)
    print("Matrix shape:", A.shape)
    print("Non-zero count:", nonzero.size)
    print("Energy (sum squares of non-zero):", energy)
    print()

lab2()

print("--- Lab9 Program 3: Monte-Carlo Estimation of Pi ---")
def lab3(N=20000):
    x = np.random.rand(N)
    y = np.random.rand(N)
    inside = np.sum(x*x + y*y <= 1.0)
    pi_est = 4.0 * inside / N
    print(f"N={N} -> pi_est={pi_est:.6f}, error={abs(pi_est-np.pi):.6f}")
    print()

lab3()

print("--- Lab9 Program 4: Stability of Random 3x3 Matrices ---")
def lab4(count=10):
    matrices = [np.random.randn(3,3) for _ in range(count)]
    for i,A in enumerate(matrices,1):
        eigs = np.linalg.eigvals(A)
        stable = np.all(np.abs(eigs) < 1)
        print(f"Matrix {i}: stable={stable}, eigs={np.round(eigs,4)}")
    print()

lab4()

print("--- Lab9 Program 5: Sliding Windows and Covariance ---")
def lab5():
    rng = np.random.default_rng(0)
    t = np.linspace(0,20,200)
    f1 = np.sin(t) + 0.1*rng.standard_normal(t.size)
    f2 = 0.8*f1 + 0.2*rng.standard_normal(t.size)
    f3 = 0.5*rng.standard_normal(t.size)
    X = np.vstack([f1,f2,f3]).T
    window = 50
    cov = np.cov(X[:window,:], rowvar=False)
    print("First window covariance (shape=", cov.shape,"):")
    print(np.round(cov,4))
    print()

lab5()

print("--- Lab9 Program 6: Mean, Variance and Median of 5000 samples ---")
def lab6():
    rng = np.random.default_rng(1)
    samples = rng.normal(loc=10.0, scale=2.5, size=5000)
    print("Mean:", np.mean(samples))
    print("Variance:", np.var(samples))
    print("Median:", np.median(samples))
    print()

lab6()

print("--- Lab9 Program 7: Pairs of Highly Correlated Features ---")
def lab7():
    rng = np.random.default_rng(0)
    n = 500
    base1 = rng.normal(size=n)
    data = np.column_stack([
        base1,
        0.9*base1 + 0.1*rng.normal(size=n),
        0.5*base1 + 0.5*rng.normal(size=n),
        rng.normal(size=n)
    ])
    cols = [f"f{i+1}" for i in range(data.shape[1])]
    df = pd.DataFrame(data, columns=cols)
    corr = df.corr()
    print(corr.round(3))
    print()

lab7()

print("--- Lab9 Program 8: Principal Component Analysis (PCA) Example ---")
def lab8():
    rng = np.random.default_rng(2)
    n_samples = 300
    t = rng.normal(size=(n_samples,3))
    W = np.array([[1.0,0.8,0.5,0.0,0.2],[0.2,0.9,0.1,0.4,0.0],[0.1,0.2,0.9,0.5,0.3]])
    X = t.dot(W) + 0.1*rng.normal(size=(n_samples, W.shape[1]))
    Xs = StandardScaler().fit_transform(X)
    pca = PCA(n_components=3)
    pca.fit(Xs)
    print("Explained variance ratios:", np.round(pca.explained_variance_ratio_,4))
    print()

lab8()

print("--- Lab9 Program 9: Most Influential Node (Graph Centrality) ---")
def lab9():
    G = nx.barabasi_albert_graph(n=50, m=2, seed=42)
    deg = nx.degree_centrality(G)
    try:
        eig = nx.eigenvector_centrality(G, max_iter=1000)
    except nx.PowerIterationFailedConvergence:
        eig = nx.eigenvector_centrality_numpy(G)
    top_eig = sorted(eig.items(), key=lambda x: x[1], reverse=True)[:5]
    print("Top 5 nodes by eigenvector centrality:")
    for node, score in top_eig:
        print(f"Node {node}: {score:.4f}")
    print()

lab9()

print("--- All programs executed. ---")
```

## 1. Random Sensor Readings and Normalization

### AIM

To generate 1000 random sensor readings using a generator expression and normalize the readings using the mean and standard deviation.

### PROGRAM

```python
import numpy as np

# Generate 1000 random numbers
data = np.fromiter((np.random.rand() for _ in range(1000)), dtype=float)

# Compute mean and standard deviation of original data
mean = np.mean(data)
std = np.std(data)

# Normalize the data
normalized = (data - mean) / std

# Print results
print("Original Mean =", mean)
print("Original Standard Deviation =", std)
print("Normalized Data =", normalized)

# Check mean and std of normalized data
print("Normalized Mean =", np.mean(normalized))
print("Normalized Standard Deviation =", np.std(normalized))
print(np.isclose(np.mean(normalized), 0))
print(round(np.mean(normalized), 10))
```

## 3. Monte-Carlo Estimation of π

### AIM

To estimate the value of π using randomly generated points and the Monte-Carlo method.

### PROGRAM

```python
import numpy as np

N = 10000

points = ((np.random.rand(), np.random.rand()) for _ in range(N))

inside = sum(1 for x, y in points if x*x + y*y <= 1)

pi_est = 4 * inside / N

print("Estimated Pi:", pi_est)
```

## 4. Stability of Random 3 × 3 Matrices

### AIM

To generate 10 random 3 × 3 matrices and retain the matrices whose eigenvalues all have magnitude less than 1.

### PROGRAM

```python
import numpy as np

# Generate 10 random 3x3 matrices
matrices = [np.random.randn(3, 3) for _ in range(10)]

# Select only stable matrices
stable = (A for A in matrices if np.all(np.abs(np.linalg.eigvals(A)) < 1))

# Print each stable matrix
for s in stable:
    print("Stable Matrix:\n", s)
```

## 8. Principal Component Analysis (PCA)

### AIM

To create a 100 × 5 dataset, calculate its covariance matrix, find the first principal component, and project the centered data onto it.

### PROGRAM

```python
import numpy as np

X = np.random.rand(100, 5)

X_centered = X - np.mean(X, axis=0)

cov = np.cov(X_centered, rowvar=False)

eigvals, eigvecs = np.linalg.eig(cov)

pc1 = eigvecs[:, np.argmax(eigvals)]

projection = X_centered @ pc1

print(projection)
```

## 9. Most Influential Node from a Dominant Eigenvector

### AIM

To generate a 6 × 6 adjacency matrix, compute its eigenvalues and eigenvectors, and identify the node with the largest component in the dominant eigenvector.

### PROGRAM

```python
import numpy as np

A = np.random.randint(0, 2, (6, 6))

eigvals, eigvecs = np.linalg.eig(A)

dominant = eigvecs[:, np.argmax(eigvals)]

node = max((i for i in range(len(dominant))),
           key=lambda i: abs(dominant[i]))

print("Most influential node:", node)
```

### 1. For $A = \begin{bmatrix}3 & 1\\5 & 2\end{bmatrix}$, compute $\det(A)$ and $A^{-1}$ by hand, then verify using NumPy.


Given,

$$
A =
\begin{bmatrix}
3 & 1\\
5 & 2
\end{bmatrix}
$$

$$
\det(A) = (3)(2) - (1)(5)
$$

$$
= 6 - 5
$$

$$
\det(A) = 1 \neq 0
$$

Therefore, inverse exists.

$$
A^{-1} = \frac{1}{\det(A)}\operatorname{adj}(A)
$$

$$
= \frac{1}{1}
\begin{bmatrix}
2 & -1\\
-5 & 3
\end{bmatrix}
$$

Therefore,

$$
A^{-1} =
\begin{bmatrix}
2 & -1\\
-5 & 3
\end{bmatrix}
$$

```python
import numpy as np

A = np.array([[3, 1],
              [5, 2]])

det_A = np.linalg.det(A)
inv_A = np.linalg.inv(A)

print("det(A) =", det_A)
print("A^(-1) =\n", inv_A)
```

### 5. Perform QR decomposition of $\begin{bmatrix}1&1\\1&-1\end{bmatrix}$ and verify $Q^TQ=I$.

Given,

$$
A=
\begin{bmatrix}
1&1\\
1&-1
\end{bmatrix}
$$

Let,

$$
a_1=
\begin{bmatrix}
1\\
1
\end{bmatrix}
$$

$$
\|a_1\|=\sqrt{1^2+1^2}=\sqrt{2}
$$

Therefore,

$$
q_1=\frac{a_1}{\|a_1\|}
=
\frac{1}{\sqrt{2}}
\begin{bmatrix}
1\\
1
\end{bmatrix}
$$

For,

$$
a_2=
\begin{bmatrix}
1\\
-1
\end{bmatrix}
$$

$$
r_{12}=q_1^Ta_2
$$

$$
=
\begin{bmatrix}
\frac{1}{\sqrt2}&\frac{1}{\sqrt2}
\end{bmatrix}
\begin{bmatrix}
1\\
-1
\end{bmatrix}
=0
$$

Therefore,

$$
u_2=a_2-r_{12}q_1=a_2
$$

$$
\|u_2\|=\sqrt{2}
$$

Therefore,

$$
q_2=
\frac{1}{\sqrt2}
\begin{bmatrix}
1\\
-1
\end{bmatrix}
$$

Hence,

$$
Q=
\begin{bmatrix}
\frac1{\sqrt2}&\frac1{\sqrt2}\\
\frac1{\sqrt2}&-\frac1{\sqrt2}
\end{bmatrix}
$$

and

$$
R=
\begin{bmatrix}
\sqrt2&0\\
0&\sqrt2
\end{bmatrix}
$$

Verification:

$$
QR=
\begin{bmatrix}
\frac1{\sqrt2}&\frac1{\sqrt2}\\
\frac1{\sqrt2}&-\frac1{\sqrt2}
\end{bmatrix}
\begin{bmatrix}
\sqrt2&0\\
0&\sqrt2
\end{bmatrix}
$$

$$
=
\begin{bmatrix}
1&1\\
1&-1
\end{bmatrix}
=A
$$

```python
import numpy as np

A = np.array([[1, 1],
              [1, -1]])

Q, R = np.linalg.qr(A)

print("Q =\n", Q)
print("R =\n", R)
print("Q.T @ Q =\n", Q.T @ Q)
```

$$

$$

### 9. Explain why $0.1+0.2==0.3$ is false and show the correct comparison.


We know,

$$
0.1+0.2=0.3
$$

But computers store floating-point numbers in binary.

In binary,

$$
0.1=0.0001100110011\ldots
$$

$$
0.2=0.0011001100110\ldots
$$

$$
0.3=0.0100110011001\ldots
$$

These values cannot be represented exactly in binary floating-point.

Therefore,

$$
0.1+0.2\approx0.30000000000000004
$$

So,

$$
0.1+0.2==0.3
$$

is False.

Correct comparison:

$$
\operatorname{isclose}(0.1+0.2,0.3)
$$

This gives True.

```python
import numpy as np

print(np.isclose(0.1 + 0.2, 0.3))
```

### 11. Compare an $O(n)$ loop with an $O(n^2)$ nested loop using operation counts.


Let,

$$
n=1000
$$

### $O(n)$ loop

$$
\sum_{i=1}^{n}1=n
$$

Therefore,

$$
O(n)\text{ operations}=1000
$$

### $O(n^2)$ nested loop

$$
\sum_{i=1}^{n}\sum_{j=1}^{n}1
$$

$$
=n\times n
$$

$$
=n^2
$$

For $n=1000$,

$$
1000\times1000=1,000,000
$$

Therefore,

$$
O(n^2)\text{ operations}=1,000,000
$$

Comparison:

$$
\frac{O(n^2)}{O(n)}
=
\frac{1,000,000}{1000}
=1000
$$

Therefore,

$$
\boxed{O(n^2)\text{ performs 1000 times more operations}}
$$

```python
#Loop
n = 1000

count1 = 0
for i in range(n):
    count1 += 1

# Nested Loop
count2 = 0
for i in range(n):
    for j in range(n):
        count2 += 1

print("O(n) operations =", count1)
print("O(n^2) operations =", count2)
```

### 12. Use bisection to solve $x^3-2=0$ on $[1,2]$ to tolerance $10^{-8}$.

Given,

$$
f(x)=x^3-2
$$

Interval:

$$
[1,2]
$$

Tolerance:

$$
10^{-8}
$$

At $x=1$,

$$
f(1)=1^3-2=-1
$$

At $x=2$,

$$
f(2)=2^3-2=6
$$

Since,

$$
f(1)<0,\qquad f(2)>0
$$

the root lies in $[1,2]$.

Bisection formula:

$$
c=\frac{a+b}{2}
$$

### Iteration 1

$$
c=\frac{1+2}{2}=1.5
$$

$$
f(1.5)=1.5^3-2
$$

$$
=3.375-2=1.375>0
$$

Therefore, root lies between:

$$
[1,1.5]
$$

### Iteration 2

$$
c=\frac{1+1.5}{2}=1.25
$$

$$
f(1.25)=1.25^3-2
$$

$$
=1.953125-2=-0.046875<0
$$

Therefore, root lies between:

$$
[1.25,1.5]
$$

### Iteration 3

$$
c=\frac{1.25+1.5}{2}=1.375
$$

$$
f(1.375)\approx0.5996>0
$$

Therefore, root lies between:

$$
[1.25,1.375]
$$

Continue the bisection process until the interval width is less than

$$
10^{-8}
$$

The root is approximately,

$$
\boxed{x\approx1.25992105}
$$

```python
def f(x):
    return x**3 - 2

a = 1
b = 2
tol = 1e-8

while (b - a) > tol:
    c = (a + b) / 2

    if f(a) * f(c) < 0:
        b = c
    else:
        a = c

print("Root =", (a + b) / 2)
print("Error =", abs(b - a))
```

### 13. Approximate $e^x$ at $x=1$ using forward differences for several $h$ values and identify a good range for $h$.

Forward difference formula:

$$
f'(x)\approx\frac{f(x+h)-f(x)}{h}
$$

Given,

$$
f(x)=e^x
$$

and

$$
x=1
$$

Therefore,

$$
f'(x)=e^x
$$

At $x=1$,

$$
f'(1)=e
$$

So,

$$
e\approx\frac{e^{1+h}-e^1}{h}
$$

For $h=0.1$:

$$
f'(1)\approx
\frac{e^{1.1}-e^1}{0.1}
$$

$$
\approx2.85884
$$

For $h=0.01$:

$$
f'(1)\approx
\frac{e^{1.01}-e^1}{0.01}
$$

$$
\approx2.73192
$$

For $h=0.001$:

$$
f'(1)\approx
\frac{e^{1.001}-e^1}{0.001}
$$

$$
\approx2.71964
$$

For $h=0.0001$:

$$
f'(1)\approx
\frac{e^{1.0001}-e^1}{0.0001}
$$

$$
\approx2.71842
$$

Exact value:

$$
e\approx2.71828
$$

As $h$ becomes smaller, the approximation gets closer to $e$.

Therefore, a good range for $h$ is approximately,

$$
\boxed{0.001\text{ to }0.0001}
$$

```python
import numpy as np

x = 1
h_values = [0.1, 0.01, 0.001, 0.0001]

for h in h_values:
    approx = (np.exp(x + h) - np.exp(x)) / h
    print("h =", h, "Approx =", approx)
```
